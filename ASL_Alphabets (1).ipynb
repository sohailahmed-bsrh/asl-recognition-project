## **1. Executive Summary**

### **1.1 Project Overview**

The **Real-Time Sign Language Keyboard System** converts American Sign Language (ASL) hand gestures into text using a webcam and a deep-learning image classification model.  
The system continuously detects gestures, predicts letters A–Z, and appends them to a text buffer, allowing users to form words and sentences like a virtual keyboard.

### **1.2 Motivation**

Sign language users often face communication barriers when interacting with people who do not understand sign language.  
By turning hand gestures into text in real time, this project aims to provide an assistive interface that bridges the gap between sign language and written communication.

### **1.3 Key Features Table**

| Feature                    | Description                                        |
|----------------------------|----------------------------------------------------|
| Gesture recognition        | Detects ASL alphabet hand signs via webcam        |
| Real-time prediction       | Processes frames continuously with low latency    |
| Text buffer keyboard       | Converts predicted letters into text strings      |
| Transfer learning model    | Uses pretrained CNN (MobileNetV2 / ResNet / etc.)|
| ASL Kaggle dataset         | Trains on labeled ASL alphabet images             |

# **24.Label Encoding**

````python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# ==========================================================
# ImageDataGenerator
# ==========================================================

train_datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.20
)

# ==========================================================
# Training Generator
# ==========================================================

train_generator = train_datagen.flow_from_directory(
    train_data_root,
    target_size=(224, 224),
    batch_size=32,
    class_mode="categorical",
    subset="training",
    shuffle=True,
    seed=42
)

# ==========================================================
# Validation Generator
# ==========================================================

validation_generator = train_datagen.flow_from_directory(
    train_data_root,
    target_size=(224, 224),
    batch_size=32,
    class_mode="categorical",
    subset="validation",
    shuffle=False,
    seed=42
)

````

# **30. MobileNetV2 Architechture**

````python
# ==========================================================
# MOBILENETV2 ARCHITECTURE
# ==========================================================

base_model = MobileNetV2(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

base_model.trainable = False

print("=" * 60)
print("MOBILENETV2 ARCHITECTURE")
print("=" * 60)

print(f"Model Name        : {base_model.name}")
print(f"Input Shape       : {base_model.input_shape}")
print(f"Output Shape      : {base_model.output_shape}")
print(f"Trainable Layers  : {base_model.trainable}")

print("=" * 60)
print("Base Model Loaded Successfully")
````

### **Interpretation**
The pretrained MobileNetV2 architecture was successfully loaded without its original classification head by setting include_top=False. The model accepts RGB images with dimensions of 224 × 224 × 3 and produces high-level feature maps of size 7 × 7 × 1280. All pretrained layers were frozen (trainable = False) to preserve the learned ImageNet features during the initial training phase. This allows the network to act as an efficient feature extractor while the newly added classification layers learn the ASL Alphabet dataset.

# **31. Model Construction**

````python
# ============================================================
# MODEL CONSTRUCTION
# ============================================================

from tensorflow.keras.models import Model
from tensorflow.keras.layers import GlobalAveragePooling2D
from tensorflow.keras.layers import Dropout
from tensorflow.keras.layers import Dense

# Freeze the base model
base_model.trainable = False

# Build custom classification head
x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dropout(0.30)(x)

output_layer = Dense(
    train_generator.num_classes,
    activation="softmax"
)(x)

# Final Model
model = Model(
    inputs=base_model.input,
    outputs=output_layer
)

print("=" * 60)
print("MODEL CONSTRUCTION")
print("=" * 60)
print(f"Input Shape      : {model.input_shape}")
print(f"Output Shape     : {model.output_shape}")
print(f"Total Layers     : {len(model.layers)}")
print(f"Output Classes   : {train_generator.num_classes}")
print("=" * 60)

print("Custom Classification Model Created Successfully")
````

### **Interpretation:**
The MobileNetV2 base model was successfully extended with custom classification layers. A Global Average Pooling layer was used to reduce feature dimensions, followed by a Dropout layer to reduce overfitting. Finally, a Dense layer with Softmax activation was added to classify images into 29 ASL alphabet classes. The model is now fully constructed and ready for compilation and training.

# **32. Model Compilation**

````python
# ============================================================
# MODEL COMPILATION
# ============================================================

from tensorflow.keras.optimizers import Adam

model.compile(
    optimizer=Adam(learning_rate=0.0001),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print("=" * 60)
print("MODEL COMPILATION")
print("=" * 60)
print("Optimizer      : Adam")
print("Learning Rate  : 0.0001")
print("Loss Function  : Categorical Crossentropy")
print("Evaluation     : Accuracy")
print("=" * 60)

print("Model Compiled Successfully")
````

### **Interpretation**
The model was successfully compiled using the Adam optimizer with a learning rate of 0.0001, which provides stable and efficient weight updates during training. Since the ASL Alphabet dataset contains multiple classes, the Categorical Crossentropy loss function was selected. Accuracy was chosen as the evaluation metric to monitor the model's classification performance throughout the training process. The model is now fully configured and ready for the training phase.

# **33.Model Training**

````python
# ============================================================
# MODEL TRAINING
# ============================================================

from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

# Early Stopping
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True,
    verbose=1
)

# Save Best Model
checkpoint = ModelCheckpoint(
    "best_asl_model.keras",
    monitor="val_accuracy",
    save_best_only=True,
    verbose=1
)

# Train Model
history = model.fit(
    train_generator,
    validation_data=validation_generator,
    epochs=20,
    callbacks=[early_stop, checkpoint]
)

print("=" * 60)
print("MODEL TRAINING COMPLETED")
print("=" * 60)

print(f"Training Epochs Completed : {len(history.history['loss'])}")
print(f"Final Training Accuracy   : {history.history['accuracy'][-1]:.4f}")
print(f"Final Validation Accuracy : {history.history['val_accuracy'][-1]:.4f}")

print("=" * 60)
````

### **Interpretation**
The MobileNetV2 model was trained for a maximum of 20 epochs using transfer learning. Early Stopping automatically terminated the training after 18 epochs because the validation accuracy stopped improving. The best-performing model was saved as best_asl_model.keras, and the weights from the best epoch (Epoch 13) were restored. The model achieved 99.19% training accuracy and 85.80% validation accuracy, indicating that it learned the training data effectively while maintaining good generalization on the validation dataset.

# **42. Text Buffer Module**

````python
# ============================================================
# STEP 9: TEXT BUFFER MODULE
# ============================================================

import time

# Store detected text
typed_text = ""

# Store previous prediction
previous_prediction = ""

# Last letter added time
last_added_time = time.time()

# Time delay between two letters (seconds)
LETTER_DELAY = 1.0

# Minimum confidence required
CONFIDENCE_THRESHOLD = 90

print("="*60)
print("TEXT BUFFER INITIALIZED SUCCESSFULLY")
print("="*60)

print("Typed Text :", typed_text)
print("Previous Prediction :", previous_prediction)
print("Letter Delay :", LETTER_DELAY, "seconds")
print("Confidence Threshold :", CONFIDENCE_THRESHOLD,"%")
````

# **42.1 -- Add Letter Fuction**

````python
# ============================================================
# ADD LETTER FUNCTION
# ============================================================

def add_letter(predicted_letter, confidence):

    global typed_text
    global previous_prediction
    global last_added_time

    current_time = time.time()

    if (
        confidence >= CONFIDENCE_THRESHOLD
        and predicted_letter != previous_prediction
        and (current_time - last_added_time) > LETTER_DELAY
    ):

        typed_text += predicted_letter

        previous_prediction = predicted_letter

        last_added_time = current_time

    return typed_text
````
